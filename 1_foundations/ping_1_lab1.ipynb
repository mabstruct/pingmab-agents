```python
# And now - the all important import statement
# If you get an import error - head over to Q5 and Q25 in the FAQ at https://edwarddonner.com/avatar
# Even for other LLM providers like Gemini, you still use this OpenAI import - see Guide 9 for why

from openai import OpenAI
```

```python
# And now we'll create an instance of the OpenAI class
# If you're not sure what it means to create an instance of a class - head over to the guides folder (guide 6)!
# If you get a NameError - head over to the guides folder (guide 6)to learn about NameErrors - always instantly fixable
# If you're not using OpenAI, you just need to slightly modify this - precise instructions are in the AI APIs guide (guide 9)

openai = OpenAI()
```

```python
# Create a list of messages in the familiar OpenAI format

messages = [{"role": "user", "content": "I am interested in using agentic solutions for API gateways. What do I have to consider when thinking about architecture in this field."}]
```

```python
messages
```

```python
# And now call it! Any problems, head to the troubleshooting guide
# This uses GPT 5.4 nano, an incredibly cheap model
# The APIs guide (guide 9) has exact instructions for using even cheaper or free alternatives to OpenAI
# If you get a NameError, head to the guides folder (guide 6) to learn about NameErrors - always instantly fixable

response = openai.chat.completions.create(model="gpt-5.5", messages=messages)
answer = response.choices[0].message.content
print(answer)
```

```python
# And now - let's ask for a question:

question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
# ask it - this uses GPT 5.4 mini, still cheap but more powerful than nano

response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
# load environment variables
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv(), override=True)
```

```python
# First create the messages:
bd_message = [{"role": "user", 
                "content": 
                """We are exploring the usage of agentic solutions for some promissing new business domains. 
Please select a list of 5 promissing business domains. Choos a random elemt out of that list. 
Respond only with the name of the chosen business domain."""}]

# Ask for a business domain:

response = openai.chat.completions.create(
    model="gpt-5.4-nano",
    messages=bd_message
)

# read out the business domain response
business_domain = response.choices[0].message.content
print(business_domain)
```

```python
# create a message to ask for a pain-point:
mess_pain_point = [
    {
        "role": "user", 
        "content": f"""We are exploring the business domain of {business_domain} for the use of Agentic AI. 
                    Please suggest a pain-point in that industry - something challenging that might be ripe for an Agentic solution. 
                    Output only the name of the business domain, the pain-point title and a short description of that paint point in two sentences.
                    """
    }
]

# Ask for a business domain:

response = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=mess_pain_point
)

# print the business domain response
pain_point = response.choices[0].message.content
print(pain_point)
```

```python
# create a message to addresses an Agentic AI solution for the given paint point
mess_agentic_solution = [
    {
        "role": "user", 
        "content": f"""We are exploring the business domain of {business_domain} for the use of Agentic AI. 
        Please suggest an Agentic AI solution for the pain-point {pain_point}.
        Output the business domain, the pain-point title and descriptoin.
        And give a description of the Agentic AI solution, including the core capabilities and benefits, the features and how it works and an overview of the  architecture and implementation approach.
        """
    }
]

# Ask for a business domain:

response = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=mess_agentic_solution
)

# print the business domain response
agentic_solution = response.choices[0].message.content
print(agentic_solution)
```

```python
	
from IPython.display import Markdown, display

display(Markdown(agentic_solution))
```
